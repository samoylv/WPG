## Integration with [phenom](https://github.com/twguest/phenom)

The python package **`phenom`** is designed to provide a simple, robust and computationally efficient method for generating representations of the complex wavefield of X-ray Free Electron Laser pulses. By making use of approximate representations of pulse wavefront and spectra, phenom allows large ensembles of photon pulses with arbitrary statistics to be generated in a truly python-ised manner.

### The **`source.py`** module:

- **`sase_pulse`**: Converts pulses generated using the **`phenom`** package into **`wpg.Wavefront`** objects. This function allows the definition of a SASE (Self-Amplified Spontaneous Emission) pulse with customizable parameters, including spatial coordinates, time array, photon energy, pulse energy, pulse duration, bandwidth, beam size, divergence, initial positions, tilt angles, domain type, and polarization. It constructs the electric field of the pulse in the specified domain and polarization and then initializes a **`wpg.Wavefront`** object with this data, preparing it for further wavefront propagation and analysis within the wpg framework.

### Changes to **`wpg.Wavefront`**

- **`__getattr__`**: Retrieves nested attributes from `params` or its `Mesh` attribute or raises an `AttributeError`.

- **`set_electric_field_representation`**: Sets the electric field representation to a specified domain using `srwlpy.SetRepresElecField`.

- **`get_pixel_size`**: Calculates the pixel size along a specified axis based on the mesh grid parameters and electric field representation.

- **`get_axis`**: Generates and returns an array of values for a specified axis based on the mesh grid parameters and electric field representation.

- **`get_profile`**: Extracts a one-dimensional line profile from the intensity distribution of the wavefront along a specified axis using a specified method.

- **`get_beam_width`**: Computes the Gaussian width of the intensity distribution along a specified axis by fitting a Gaussian function to the extracted profile.

- **`get_efield`**: Converts electric field data to a complex representation based on the specified polarization ('horizontal', 'vertical', or 'circular').

- **`wavelength`**: Computes and returns the wavelength based on the photon energy using the formula \(\lambda = \frac{hc}{E}\).

- **`divergence`**: Calculates and returns the beam divergence width in reciprocal space for the `qx` and `qy` axes.

- **`pulse_duration`**: Computes and returns the Gaussian width of the pulse duration in the time domain.
